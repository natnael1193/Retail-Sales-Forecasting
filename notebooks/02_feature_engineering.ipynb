## Feature Engineering

```python
# Import the data from 01_eda.ipynb
%store -r df_store
```

```python
# Add date features
df_store["month"] = df_store["date"].dt.month
df_store["dayofweek"] = df_store["date"].dt.dayofweek
df_store["weekofyear"] = df_store["date"].dt.isocalendar().week.astype(int)
df_store["year"] = df_store["date"].dt.year
```

```python
# Create LAG features
# A lag feature is: A previous value of the target variable used as an input feature


# sort values by store and date
df_store = df_store.sort_values(by=["store_nbr", "date"])
print(df_store.head())
```

```python
# Create the LAGs

df_store["lag_1"] = df_store.groupby("store_nbr")["sales"].shift(1)
df_store["lag_7"] = df_store.groupby("store_nbr")["sales"].shift(7)
df_store["lag_30"] = df_store.groupby("store_nbr")["sales"].shift(30)

print(df_store.head(35))
```

```python
# Applying Window based Calculation (Rolling)

df_store["rolling_7"] = (
    df_store.groupby("store_nbr")["sales"]
            .shift(1)
            .rolling(7)
            .mean()
)

print("", df_store.head(10))

df_store["rolling_30"] = (
    df_store.groupby("store_nbr")["sales"]
            .shift(1)
            .rolling(30)
            .mean()
)

print( df_store.head(35))
```

```python
# Drop NA 

df_store = df_store.dropna()

print("Shape after dropping NA:", df_store.shape)
print('\n')
print(df_store.head())
```
