```python
# Generate a professional PDF report for the Retail Sales Forecasting Project

from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, ListFlowable, ListItem
from reportlab.lib.styles import ParagraphStyle, getSampleStyleSheet
from reportlab.lib import colors
from reportlab.lib.units import inch
from reportlab.platypus import HRFlowable
from reportlab.pdfbase.ttfonts import TTFont
from reportlab.pdfbase import pdfmetrics
import os

# File path - save to the project directory
file_path = os.path.join("..", "reports", "Retail_Sales_Forecasting_Project_Summary_Report.pdf")
doc = SimpleDocTemplate(file_path)
elements = []

styles = getSampleStyleSheet()
title_style = styles["Heading1"]
section_style = styles["Heading2"]
normal_style = styles["BodyText"]

# Title
elements.append(Paragraph("Retail Sales Forecasting for Multi-Store Retailer", title_style))
elements.append(Spacer(1, 0.3 * inch))

# Executive Summary
elements.append(Paragraph("Executive Summary", section_style))
elements.append(Spacer(1, 0.2 * inch))
elements.append(Paragraph(
    "This project develops a multi-store retail demand forecasting system designed to improve inventory planning and reduce forecast error across stores. "
    "Using machine learning techniques and time-series feature engineering, the final XGBoost model achieved a 57% reduction in RMSE compared to a naive baseline and reached a MAPE of 9.75%.",
    normal_style
))
elements.append(Spacer(1, 0.3 * inch))

# Business Objective
elements.append(Paragraph("Business Objective", section_style))
elements.append(Spacer(1, 0.2 * inch))
elements.append(Paragraph(
    "The goal of this project is to accurately forecast daily sales per store to support inventory optimization, reduce stockouts and overstock, and improve operational efficiency in a retail environment.",
    normal_style
))
elements.append(Spacer(1, 0.3 * inch))

# Data Description
elements.append(Paragraph("Dataset Description", section_style))
elements.append(Spacer(1, 0.2 * inch))
data_points = [
    "Date: Daily timestamp of sales",
    "Store Number: Unique store identifier",
    "Product Family: Product category",
    "Sales: Target variable (daily sales)",
    "On Promotion: Number of items under promotion"
]

elements.append(ListFlowable(
    [ListItem(Paragraph(point, normal_style)) for point in data_points],
    bulletType='bullet'
))
elements.append(Spacer(1, 0.3 * inch))

# Methodology
elements.append(Paragraph("Methodology", section_style))
elements.append(Spacer(1, 0.2 * inch))
method_points = [
    "Aggregated sales at store-day level",
    "Created time-based features (month, day of week, week of year)",
    "Engineered lag features (lag_1, lag_7, lag_30)",
    "Created rolling averages (7-day and 30-day windows)",
    "Time-based train/test split (last 3 months as test set)"
]

elements.append(ListFlowable(
    [ListItem(Paragraph(point, normal_style)) for point in method_points],
    bulletType='bullet'
))
elements.append(Spacer(1, 0.3 * inch))

# Models
elements.append(Paragraph("Models Implemented", section_style))
elements.append(Spacer(1, 0.2 * inch))
model_points = [
    "Naive Baseline (Lag-1 Forecast)",
    "Random Forest Regressor",
    "XGBoost Regressor"
]

elements.append(ListFlowable(
    [ListItem(Paragraph(point, normal_style)) for point in model_points],
    bulletType='bullet'
))
elements.append(Spacer(1, 0.3 * inch))

# Results
elements.append(Paragraph("Model Performance", section_style))
elements.append(Spacer(1, 0.2 * inch))
results_points = [
    "Baseline RMSE: 5348",
    "Random Forest RMSE: 2485",
    "XGBoost RMSE: 2278",
    "MAPE: 9.75%",
    "RMSE Reduction vs Baseline: 57%"
]

elements.append(ListFlowable(
    [ListItem(Paragraph(point, normal_style)) for point in results_points],
    bulletType='bullet'
))
elements.append(Spacer(1, 0.3 * inch))

# Business Impact
elements.append(Paragraph("Business Impact", section_style))
elements.append(Spacer(1, 0.2 * inch))
elements.append(Paragraph(
    "By reducing forecast error by 57%, the model significantly improves demand planning accuracy. "
    "Lower forecasting error directly contributes to reduced safety stock requirements, fewer stockouts, "
    "and improved supply chain efficiency. Achieving sub-10% MAPE demonstrates production-ready retail forecasting capability.",
    normal_style
))
elements.append(Spacer(1, 0.3 * inch))

# Conclusion
elements.append(Paragraph("Conclusion", section_style))
elements.append(Spacer(1, 0.2 * inch))
elements.append(Paragraph(
    "This project demonstrates the application of machine learning for retail demand forecasting. "
    "Through feature engineering and advanced modeling techniques, the system delivers strong predictive performance "
    "and provides measurable business value for inventory optimization and operational planning.",
    normal_style
))

# Build document
doc.build(elements)

print(f"PDF report saved to: {file_path}")
print(f"File exists: {os.path.exists(file_path)}")
```

```python
# Generate an Expanded Technical PDF Report (15+ pages style content)

from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, PageBreak, ListFlowable, ListItem
from reportlab.lib.styles import getSampleStyleSheet
from reportlab.lib.units import inch
from reportlab.platypus import HRFlowable

file_path_detailed = "../reports/Retail_Sales_Forecasting_Full_Report.pdf"
doc = SimpleDocTemplate(file_path_detailed)
elements = []
styles = getSampleStyleSheet()

def add_section(title, content_paragraphs):
    elements.append(Paragraph(title, styles["Heading1"]))
    elements.append(Spacer(1, 0.3 * inch))
    for para in content_paragraphs:
        elements.append(Paragraph(para, styles["BodyText"]))
        elements.append(Spacer(1, 0.2 * inch))
    elements.append(PageBreak())

# Sections
add_section("1. Introduction", [
    "Retail demand forecasting plays a critical role in inventory optimization and supply chain efficiency.",
    "This project builds a multi-store sales forecasting system using machine learning techniques.",
    "The objective is to reduce forecast error and provide measurable business impact."
])

add_section("2. Business Problem Definition", [
    "Fashion retail operates in a fast-changing demand environment.",
    "Inaccurate forecasts lead to stockouts or excess inventory.",
    "The goal is to predict daily store-level sales to improve planning accuracy."
])

add_section("3. Dataset Description", [
    "The dataset includes daily sales data across multiple stores.",
    "Key variables include date, store number, product family, sales, and promotion count.",
    "Sales were aggregated at the store-day level for forecasting."
])

add_section("4. Exploratory Data Analysis", [
    "EDA revealed strong weekly seasonality and trend patterns.",
    "Promotional activity showed clear correlation with sales spikes.",
    "Store-level variability justified multi-store modeling."
])

add_section("5. Feature Engineering", [
    "Time-based features: month, day of week, week of year, year.",
    "Lag features: lag_1, lag_7, lag_30 to capture temporal dependency.",
    "Rolling averages: 7-day and 30-day windows to capture trends."
])

add_section("6. Modeling Strategy", [
    "A time-based train-test split was used to avoid leakage.",
    "Baseline model: naive lag-1 forecast.",
    "Machine learning models: Random Forest and XGBoost."
])

add_section("7. Model Evaluation", [
    "Baseline RMSE: 5348.",
    "Random Forest RMSE: 2485.",
    "XGBoost RMSE: 2278.",
    "MAPE: 9.75%.",
    "XGBoost reduced RMSE by 57% compared to baseline."
])

add_section("8. Business Impact Analysis", [
    "Sub-10% MAPE indicates production-level forecasting capability.",
    "Improved forecast accuracy reduces safety stock requirements.",
    "Better demand prediction enhances operational efficiency."
])

add_section("9. Feature Importance Insights", [
    "Lag features were the strongest predictors, confirming temporal dependence.",
    "Weekly seasonality played a major role in demand fluctuations.",
    "Promotions contributed significantly to sales variability."
])

add_section("10. Conclusion & Future Work", [
    "The forecasting pipeline demonstrates strong predictive performance.",
    "Future work could include LSTM models and hierarchical forecasting.",
    "Integration into inventory optimization systems would enhance impact."
])

doc.build(elements)

file_path_detailed
```
